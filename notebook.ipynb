## 2. Data Understanding & Preparation

### Dataset Source
The dataset used is a **real-world retail transaction dataset**
from the UCI Machine Learning Repository.

Transaction-level data was aggregated to generate
daily total sales values.

### Dataset Description
- **Date:** Daily sales date
- **Sales:** Total daily revenue


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("daily_retail_sales.csv")

df.head()
```

### Data Cleaning & Feature Engineering

The following steps were performed:
- Converted date column to datetime format
- Sorted records chronologically
- Created time-based features
- Generated lag and rolling average features
- Removed missing values


```python
df["Date"] = pd.to_datetime(df["Date"])
df.sort_values("Date", inplace=True)

df["day"] = df["Date"].dt.day
df["month"] = df["Date"].dt.month
df["year"] = df["Date"].dt.year
df["day_of_week"] = df["Date"].dt.dayofweek

df["lag_1"] = df["Sales"].shift(1)
df["rolling_avg"] = df["Sales"].rolling(7).mean()

df.dropna(inplace=True)

df.head()
```

## 3. Model / System Design

### AI Techniques Used
- Linear Regression
- Random Forest Regressor

### Pipeline
1. Data preprocessing
2. Feature engineering
3. Model training
4. Model evaluation
5. Forecasting
6. Business insight generation

### Justification
Multiple models are used to compare performance and robustness
on time-series data.


```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

X = df[["day", "month", "year", "day_of_week", "lag_1", "rolling_avg"]]
y = df["Sales"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

lr = LinearRegression()
rf = RandomForestRegressor(n_estimators=100, random_state=42)

lr.fit(X_train, y_train)
rf.fit(X_train, y_train)
```

## 5. Evaluation & Analysis

RMSE (Root Mean Squared Error) is used to evaluate model performance.


```python
from sklearn.metrics import mean_squared_error
import numpy as np

lr_mse = mean_squared_error(y_test, lr.predict(X_test))
rf_mse = mean_squared_error(y_test, rf.predict(X_test))

lr_rmse = np.sqrt(lr_mse)
rf_rmse = np.sqrt(rf_mse)

print("Linear Regression RMSE:", lr_rmse)
print("Random Forest RMSE:", rf_rmse)
```
